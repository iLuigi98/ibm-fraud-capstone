```python
import numpy as np
import pandas as pd

# --- derive DOW + hour + cyclic hour ---
ts = pd.to_datetime(transactions["Timestamp"])
transactions["dow"]  = ts.dt.dayofweek.astype("int8")   # 0..6
transactions["hour"] = ts.dt.hour.astype("int8")        # 0..23
transactions["hour_sin"] = np.sin(2*np.pi*transactions["hour"]/24).astype("float32")
transactions["hour_cos"] = np.cos(2*np.pi*transactions["hour"]/24).astype("float32")

# --- pick sender/receiver columns (IBM AML common schema) ---
def pick_col(cands):
    for c in cands:
        if c in transactions.columns:
            return c
    raise KeyError(f"Missing columns. Tried: {cands}. Found: {list(transactions.columns)}")

from_bank = pick_col(["From Bank", "from_bank"])
from_acct = pick_col(["Account", "from_account"])
to_bank   = pick_col(["To Bank", "to_bank"])
to_acct   = pick_col(["Account.1", "To Account", "to_account", "account_1"])

# --- map accounts to node ids (USE 1D keys; factorize needs 1D) ---
src_key = transactions[from_bank].astype(str) + "|" + transactions[from_acct].astype(str)
dst_key = transactions[to_bank].astype(str)   + "|" + transactions[to_acct].astype(str)

all_keys = pd.concat([src_key, dst_key], ignore_index=True)
codes, uniques = pd.factorize(all_keys, sort=False)

E = len(src_key)
src = codes[:E].astype("int64")
dst = codes[E:].astype("int64")
num_nodes = len(uniques)

print("num_nodes:", num_nodes, "num_edges:", E)

# --- edge_attr: Amount_Log + t_sec + hour cyc + dow one-hot ---
dow_oh = np.eye(7, dtype="float32")[transactions["dow"].values]  # (E,7)

edge_attr = np.column_stack([
    transactions["Amount_Log"].values.astype("float32"),
    transactions["t_sec"].values.astype("float32"),
    transactions["hour_sin"].values.astype("float32"),
    transactions["hour_cos"].values.astype("float32"),
    dow_oh
]).astype("float32")

y_edge = transactions["Is Laundering"].values.astype("int64")  # (E,)

print("edge_attr:", edge_attr.shape, "y_edge:", y_edge.shape)
```

```python
import torch
from torch_geometric.data import Data
from torch_geometric.loader import LinkNeighborLoader

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("device:", device)

edge_index = torch.tensor(np.vstack([src, dst]), dtype=torch.long)  # [2, E]
edge_attr_t = torch.tensor(edge_attr, dtype=torch.float32)          # [E, 11]
y_edge_t    = torch.tensor(y_edge, dtype=torch.long)                # [E]

data = Data(num_nodes=num_nodes, edge_index=edge_index, edge_attr=edge_attr_t)

edge_ids = torch.arange(edge_index.size(1), dtype=torch.long)

# quick stratified-ish split (random). We'll do time split later if needed.
perm = torch.randperm(edge_ids.numel())
n_train = int(0.80 * perm.numel())
train_eids = edge_ids[perm[:n_train]]
val_eids   = edge_ids[perm[n_train:]]

train_loader = LinkNeighborLoader(
    data,
    edge_label_index=edge_index[:, train_eids],
    edge_label=y_edge_t[train_eids],
    num_neighbors=[15, 10],
    batch_size=4096,
    shuffle=True
)

val_loader = LinkNeighborLoader(
    data,
    edge_label_index=edge_index[:, val_eids],
    edge_label=y_edge_t[val_eids],
    num_neighbors=[15, 10],
    batch_size=4096,
    shuffle=False
)

print("train batches:", len(train_loader), "val batches:", len(val_loader))
```

```python
import torch.nn as nn
import torch.nn.functional as F
from torch_geometric.nn import TransformerConv

class EdgeAwareGNN(nn.Module):
    def __init__(self, num_nodes, hidden=128, edge_dim=11):
        super().__init__()
        self.emb = nn.Embedding(num_nodes, hidden)  # no node feats -> learnable
        self.conv1 = TransformerConv(hidden, hidden, edge_dim=edge_dim)
        self.conv2 = TransformerConv(hidden, hidden, edge_dim=edge_dim)
        self.mlp = nn.Sequential(
            nn.Linear(hidden*4, 256),
            nn.ReLU(),
            nn.Linear(256, 2)
        )

    def forward(self, batch):
        x = self.emb(batch.n_id)  # node embeddings for sampled nodes
        x = F.relu(self.conv1(x, batch.edge_index, batch.edge_attr))
        x = self.conv2(x, batch.edge_index, batch.edge_attr)

        src_i = batch.edge_label_index[0]
        dst_i = batch.edge_label_index[1]
        hs, hd = x[src_i], x[dst_i]
        feat = torch.cat([hs, hd, hs*hd, (hs-hd).abs()], dim=1)
        return self.mlp(feat)

model = EdgeAwareGNN(num_nodes=num_nodes, hidden=128, edge_dim=edge_attr_t.size(1)).to(device)
opt = torch.optim.AdamW(model.parameters(), lr=2e-3, weight_decay=1e-4)

def run_epoch(loader, train=True):
    model.train(train)
    total_loss, total, correct = 0.0, 0, 0
    for batch in loader:
        batch = batch.to(device)
        logits = model(batch)
        loss = F.cross_entropy(logits, batch.edge_label)
        if train:
            opt.zero_grad(set_to_none=True)
            loss.backward()
            opt.step()
        total_loss += float(loss) * batch.edge_label.size(0)
        pred = logits.argmax(dim=1)
        correct += int((pred == batch.edge_label).sum())
        total += int(batch.edge_label.size(0))
    return total_loss / max(total,1), correct / max(total,1)

for epoch in range(1, 4):
    tr_loss, tr_acc = run_epoch(train_loader, train=True)
    va_loss, va_acc = run_epoch(val_loader, train=False)
    print(f"epoch {epoch} | train loss {tr_loss:.4f} acc {tr_acc:.4f} | val loss {va_loss:.4f} acc {va_acc:.4f}")
```

```python
from sklearn.metrics import average_precision_score, roc_auc_score
import numpy as np
import torch

model.eval()
ys, ps = [], []

with torch.no_grad():
    for i, batch in enumerate(val_loader):
        batch = batch.to(device)
        logits = model(batch)
        prob1 = torch.softmax(logits, dim=1)[:, 1].detach().cpu().numpy()
        ytrue = batch.edge_label.detach().cpu().numpy()
        ys.append(ytrue)
        ps.append(prob1)
        if i >= 200:   # limit batches for speed
            break

y = np.concatenate(ys)
p = np.concatenate(ps)

print("Pos rate:", y.mean())
print("PR-AUC:", average_precision_score(y, p))
print("ROC-AUC:", roc_auc_score(y, p))
```

```python
import torch
import torch.nn.functional as F

def encode_batch(model, batch):
    # returns node embeddings for the sampled nodes in this batch
    x = model.emb(batch.n_id)
    x = F.relu(model.conv1(x, batch.edge_index, batch.edge_attr))
    x = model.conv2(x, batch.edge_index, batch.edge_attr)
    return x
```

```python
import os, numpy as np, torch
from torch_geometric.loader import NeighborLoader

save_dir = "/content/drive/MyDrive/IBM_AML_CACHE/LI_Medium"
os.makedirs(save_dir, exist_ok=True)

hidden = 128  # MUST match your model hidden size
emb_path = f"{save_dir}/node_emb_float32.memmap"

# create memmap on disk: (num_nodes, hidden)
H = np.memmap(emb_path, dtype="float32", mode="w+", shape=(num_nodes, hidden))

# NeighborLoader for node-wise inference
infer_loader = NeighborLoader(
    data,
    input_nodes=torch.arange(num_nodes),
    num_neighbors=[10, 5],   # inference sampling (faster)
    batch_size=65536,
    shuffle=False,
    num_workers=2,
    persistent_workers=True
)

model.eval()
with torch.no_grad():
    for batch in infer_loader:
        batch = batch.to(device)
        z = encode_batch(model, batch).detach().cpu().numpy()  # (batch_size, hidden)
        H[batch.n_id.cpu().numpy()] = z

H.flush()
print("Saved node embeddings to:", emb_path, "shape:", (num_nodes, hidden))
```

```python
import numpy as np
import pandas as pd

# assumes you still have transactions["Timestamp"] and y_edge in memory
ts = pd.to_datetime(transactions["Timestamp"])
order = np.argsort(ts.values)

cut = int(0.8 * len(order))
train_idx = order[:cut]   # earlier 80%
test_idx  = order[cut:]   # later 20%

print("Train pos rate:", y_edge[train_idx].mean())
print("Test  pos rate:", y_edge[test_idx].mean())
```

```python
import xgboost as xgb
from sklearn.metrics import average_precision_score, roc_auc_score

# load your saved embeddings memmap
save_dir = "/content/drive/MyDrive/IBM_AML_CACHE/LI_Medium"
H = np.memmap(f"{save_dir}/node_emb_float32.memmap", dtype="float32", mode="r", shape=(num_nodes, 128))

rng = np.random.default_rng(42)

# sample train edges (balanced)
pos_tr = train_idx[y_edge[train_idx] == 1]
neg_tr = train_idx[y_edge[train_idx] == 0]
neg_tr_s = rng.choice(neg_tr, size=min(len(neg_tr), len(pos_tr)*10), replace=False)

train_sel = np.concatenate([pos_tr, neg_tr_s])
rng.shuffle(train_sel)

# sample test edges (keep more to be robust, but cap for speed)
test_sel = test_idx
if len(test_sel) > 2_000_000:
    test_sel = rng.choice(test_sel, size=2_000_000, replace=False)

def build_X(sel):
    s = src[sel]; d = dst[sel]
    hs = H[s]; hd = H[d]
    diff = np.abs(hs - hd)
    had  = hs * hd
    return np.concatenate([hs, hd, diff, had, edge_attr[sel]], axis=1).astype("float32")

X_tr = build_X(train_sel)
y_tr = y_edge[train_sel].astype("int32")

X_te = build_X(test_sel)
y_te = y_edge[test_sel].astype("int32")

clf = xgb.XGBClassifier(
    n_estimators=500,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    reg_lambda=1.0,
    tree_method="hist",
    eval_metric="aucpr"
)
clf.fit(X_tr, y_tr)

p = clf.predict_proba(X_te)[:, 1]
print("PR-AUC:", average_precision_score(y_te, p))
print("ROC-AUC:", roc_auc_score(y_te, p))
print("Test pos rate:", y_te.mean(), "n_test:", len(y_te))
```
